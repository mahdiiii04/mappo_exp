```python
from torchrl.envs.libs.pettingzoo import PettingZooEnv
from torchrl.envs import ParallelEnv
```

```python
def make_env():
    return PettingZooEnv(
        task="simple_spread_v3",
        parallel=True,
        continuous_actions=False,
        seed=0,
        max_cycles=50,
        device="cpu"
    )

env = ParallelEnv(
    num_workers=8,
    create_env_fn=make_env,
)
```
